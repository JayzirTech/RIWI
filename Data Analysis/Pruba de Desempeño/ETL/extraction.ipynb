```python

df_address = (
    df_customer_master[[
                        'customer_city',
                        'customer_state',
                        'customer_country',
                        'region',
                        'customer_postal_code']]
)

df_address = df_address.drop_duplicates().reset_index(drop=True).reset_index()

df_address = df_address.rename(columns={'index': 'address_id'})

df_customer_master = df_customer_master.merge(
    df_address,
    on=['customer_city', 'customer_state', 'customer_country', 'region', 'customer_postal_code'],
    how='left'
)

df_customer_master = (
    df_customer_master[['customer_id',
                        'customer_name',
                        'customer_age',
                        'gender',
                        'customer_segment',
                        'customer_acquisition_cost',
                        'address_id',
                        ]]
)
```

```python
'''df_address
df_customer_master
df_product_catalog
df_returns
df_ecommerce_sales'''

from sqlalchemy import create_engine

# 1. Configura los datos de conexión a tu base de datos PostgreSQL
# Estructura: postgresql+psycopg2://usuario:contraseña@host:puerto/nombre_base_datos
user = os.getenv('DB_USER')
password = os.getenv('DB_PASSWORD')
host = os.getenv('DB_HOST')
port = os.getenv('DB_PORT')
db_name = os.getenv('DB_NAME')

# 2. Crear el motor (engine) de conexión
engine = create_engine(f'postgresql+psycopg2://{user}:{password}@{host}:{port}/{db_name}')

# 3. Diccionario con los DataFrames y los nombres que tendrán las tablas en PostgreSQL
tables_to_export = {
    'dim_address': df_address,
    'dim_customer_master': df_customer_master,
    'dim_product_catalog': df_product_catalog,
    'dim_order_items': df_order_items,
    'fact_returns' : df_returns,
    'fact_ecommerce_sales' : df_ecommerce_sales,
    'df_main' : df_main

}

# 4. Iterar y enviar cada DataFrame a su respectiva tabla en Postgres
for table_name, df in tables_to_export.items():
  try:
    df.to_sql(
        name=table_name,
        con=engine,
        if_exists='replace',  # Opciones: 'fail', 'replace' (reemplaza si existe), 'append' (agrega filas)
        index=False,  # Para que no guarde el índice numérico de Pandas como una columna extra
    )
    print(
        f'¡Éxito! DataFrame exportado correctamente a la tabla: {table_name} a {db_name.upper()}'
    )
  except Exception as e:
    print(f'Error al exportar {table_name}: {e}')
```
