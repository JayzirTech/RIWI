```python
import kagglehub
import os

path = kagglehub.dataset_download(
    "umichigan/world-religions"
)

print("Ruta:", path)

files = os.listdir(path)
print(f'\n"Archivos:", {files}')
```

```python
import pandas as pd

df_global = pd.read_csv(os.path.join(path, "global.csv"))

df_regional = pd.read_csv(os.path.join(path, "regional.csv"))

df_national = pd.read_csv(os.path.join(path, "national.csv"))
```
